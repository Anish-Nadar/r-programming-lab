## Part 1 — Data Frames: Load the `painters` dataset

```r
# MASS contains the painters data frame used in the NPTEL notes
library(MASS)

# Display the data frame
painters
```

## Part 2 — Summary of a categorical variable

```r
# Frequency table of the School variable
summary(painters$School)
```

## Part 3 — Attach a data frame and access variables directly

```r
# Attach the data frame
attach(painters)

# Now the column names can be used directly
summary(School)
```

## Part 4 — Summary of a numeric variable

```r
# Summary statistics for Composition
summary(Composition)
```

## Part 5 — Detach the data frame

```r
# Remove the attached data frame
detach(painters)

# Use painters$School after detach()
summary(painters$School)
```

## Part 6 — Subset rows using `subset()`

```r
# Select painters belonging to School F
subset(painters, School == "F")
```

## Part 7 — Equivalent row-subsetting method

```r
# Same result using logical indexing
painters[painters[["School"]] == "F", ]
```

## Part 8 — Subset using a numeric condition

```r
# Select rows where Composition is 6 or less
subset(painters, Composition <= 6)
```

## Part 9 — Remove unwanted columns while subsetting

```r
# For School F, remove Colour (column 3) and School (column 5)
subset(painters, School == "F", select = c(-3, -5))
```

## Part 10 — Split the data frame by School

```r
# Split painters into separate data frames according to School
splitted <- split(painters, painters$School)

# Display all groups
splitted
```

## Part 11 — Check one split object

```r
# splitted$A is itself a data frame
is.data.frame(splitted$A)

# Display School A
splitted$A
```

## Part 13 — Create a small Excel file for a fully executable demo

The NPTEL notes demonstrate `read_excel()` using an Excel file. To keep this notebook executable in Colab, this cell creates a small sample `.xlsx` file first.


```r
# Install writexl only if needed
if (!requireNamespace("writexl", quietly = TRUE)) {
  install.packages("writexl", repos = "https://cloud.r-project.org")
}
library(writexl)

# Create the sample data shown in the NPTEL notes
spexcel <- data.frame(
  `Variable 1` = 1:5,
  `Variable 2` = c(10, 20, 30, 40, 50),
  `Variable 3` = c(100, 200, 300, 400, 500)
)

# Create a second sheet
spexcel2 <- data.frame(
  `Variable 4` = 6:10,
  `Variable 5` = c(110, 120, 130, 140, 150),
  `Variable 6` = c(110, 210, 310, 410, 510)
)

write_xlsx(
  list(Sheet1 = spexcel, Sheet2 = spexcel2),
  path = "spexcel.xlsx"
)

list.files(pattern = "spexcel.xlsx")
```

```r
barplot(
  table(direction),
  col = c("red", "green", "blue"),
  main = "Directions of food delivery",
  legend.text = c("dir1", "dir2", "dir3")
)
```

## Part 44 — Bar plot with subtitle

```r
barplot(
  table(direction),
  col = c("red", "green", "blue"),
  main = "Directions of food delivery",
  legend.text = c("dir1", "dir2", "dir3"),
  sub = "Three directions"
)
```

## Part 45 — Final bar plot with labels

```r
barplot(
  table(direction),
  col = c("red", "green", "blue"),
  main = "Directions of food delivery",
  legend.text = c("dir1", "dir2", "dir3"),
  sub = "Three directions",
  xlab = "Food Delivery Directions",
  ylab = "Number of Deliveries"
)
```
